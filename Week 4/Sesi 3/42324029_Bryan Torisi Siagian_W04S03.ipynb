# 1. Import Library dan Load Dataset Iris
Pada cell ini dilakukan impor pustaka numpy, pandas, serta pengunduhan dataset Iris dari sklearn.datasets.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-learn
iris = load_iris()
```

# 2. Konversi Data ke Pandas DataFrame
Dataset Iris dikonversi ke dalam DataFrame Pandas. Dibuat kolom target Species, lalu dipisahkan antara variabel fitur (X) dan variabel label/target (y).

```python
# Konversi data menjadi DataFrame 
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['Species'] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

# 3. Menampilkan Label Target (y)
Cell ini menampilkan isi dari variabel y yang memuat nama-nama spesies dari bunga Iris (setosa, versicolor, virginica).

```python
y
```

# 4. Label Encoding Target Iris
Mengubah label target y yang berbentuk string/teks menjadi format numerik (0, 1, 2) menggunakan LabelEncoder.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

# 5. Split Dataset (Train-Test Split)
Membagi dataset menjadi data latih (80%) dan data uji (20%) menggunakan train_test_split dengan random_state=0.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

# 6. Pelatihan Model Gaussian Naïve Bayes
Membuat dan melatih model GaussianNB menggunakan data latih X_train dan y_train, karena fitur-fitur dataset Iris bersifat numerik kontinu.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

# 7. Prediksi Data Uji (Testing)
Melakukan prediksi label kelas pada data uji X_test menggunakan model Gaussian Naïve Bayes yang telah dilatih.

```python
result = NB_model.predict(X_test)
result
```

# 8. Evaluasi Akurasi Model
Menghitung persen akurasi dari hasil prediksi data uji menggunakan accuracy_score.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

# 9. Prediksi Data Baru (Pengujian Awal dengan Array)
Mencoba memprediksi data baru [[5.1, 3.5, 1.4, 0.2]] dalam bentuk list/array standar.

```python
# Prediksi inputan data baru
# input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah: ', prediction_label)
```

# 10. Prediksi Data Baru Menggunakan DataFrame (Solusi Warning)
Mengubah input data baru menjadi DataFrame Pandas beserta nama kolomnya agar konsisten dengan fitur yang dilatih pada model, sehingga tidak memunculkan *warning*.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah: ', prediction_label)
```

# Tugas Prediksi Data Baru
Melakukan prediksi spesies bunga Iris untuk data input baru:
* input_data = [7.3, 3.8, 4.6, 2.3]

```python
input_tugas = pd.DataFrame(np.array([[7.3, 3.8, 4.6, 2.3]]), columns=iris.feature_names)

result_tugas = NB_model.predict(input_tugas)
label_tugas = lab.inverse_transform(result_tugas)

print('Hasil prediksi tugas untuk [7.3, 3.8, 4.6, 2.3] adalah: ', label_tugas[0])
```
